## 🔐 ¿Qué tan segura es tu contraseña?

Una contraseña segura generalmente cumple con estos criterios:

| Criterio     | Recomendación                                     |
| ------------ | ------------------------------------------------- |
| Longitud     | Mínimo 12 caracteres                              |
| Complejidad  | Letras mayúsculas, minúsculas, números y símbolos |
| Unicidad     | No debe reutilizarse en diferentes plataformas    |
| Aleatoriedad | Evita palabras del diccionario, nombres o fechas  |

**Ejemplo seguro:** `B#7gYp!3k&zL`
**Ejemplo débil:** `123456`, `password`, `juan2024`

**Herramientas para evaluarla:**

* [https://haveibeenpwned.com/Passwords](https://haveibeenpwned.com/Passwords)
* [https://password.kaspersky.com](https://password.kaspersky.com)
* [https://howsecureismypassword.net](https://howsecureismypassword.net)

---

## ⚠️ Implicaciones del mal uso de políticas de contraseña

1. ### **Acceso no autorizado a sistemas**

   Si una contraseña es predecible o débil, un atacante puede adivinarla o usar técnicas como **fuerza bruta** o **ataques de diccionario** para obtener acceso.

2. ### **Robo de identidad**

   Una vez dentro, un atacante puede hacerse pasar por ti, realizar fraudes, enviar mensajes desde tu cuenta o acceder a tus datos personales.

3. ### **Uso de contraseñas compartidas**

   Compartir contraseñas o usar la misma en varias cuentas puede provocar que una sola filtración comprometa muchos sistemas.

4. ### **Phishing más efectivo**

   Si los usuarios no cambian frecuentemente sus contraseñas o usan las mismas en todo lado, los ataques por ingeniería social o phishing se vuelven más peligrosos.

5. ### **Incumplimiento normativo**

   Muchas regulaciones (como **GDPR**, **ISO 27001**, **HIPAA**) exigen políticas de seguridad robustas, incluida la gestión adecuada de contraseñas. El mal uso puede generar multas y pérdidas reputacionales.

6. ### **Pérdida de confianza**

   Si un sistema o empresa no aplica políticas de contraseñas seguras y sufre un incidente, sus clientes pueden perder la confianza, afectando la reputación de la marca o entidad.

---

## ✅ Buenas prácticas en el manejo de contraseñas

* Usa un **gestor de contraseñas** como Bitwarden, KeePass o LastPass.
* Cambia tus contraseñas si sospechas de un incidente.
* Habilita **doble factor de autenticación (2FA)** siempre que sea posible.
* Nunca compartas tus contraseñas ni las escribas en lugares visibles.
