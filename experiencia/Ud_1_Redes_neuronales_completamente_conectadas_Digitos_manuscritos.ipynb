```python
# Utiliza la función load_data() del módulo mnist de Keras para cargar el conjunto de
# datos MNIST, que contiene imágenes de dígitos escritos a mano y sus etiquetas asociadas

import keras
# Cargar el conjunto de datos MNIST
mnist = keras.datasets.mnist

#cargue la division de entrenamiento y prueba del conjunto de datos MNIST
((training_images, training_labels), (test_images, test_labels)) = mnist.load_data()
```

```python
# Visualización de una muestra: Selecciona un índice específico (index) dentro del
#conjunto de entrenamiento para visualizar una muestra de imagen y su etiqueta
#correspondiente. Imprime la etiqueta y muestra la imagen utilizando plt.imshow() de
#Matplotlib.

import numpy as np
import matplotlib.pyplot as plt

#poner aqui entre 0 y 59999

index = 1

#imprime la etiqueta y la imagen
np.set_printoptions(linewidth=320)
print(f'Label: {training_labels[index]}')
print (f'Image:\n {training_images[index]}')
#visualiza la imagen
plt.imshow(training_images[index])
```

```python
#normalizar los valores de pixeles del tren y probar las imagenes
training_images = training_images / 255.0
test_images = test_images / 255.0
```

```python
# construir el modelo de clasificacion

model = keras.models.Sequential([
    keras.layers.Flatten(input_shape= (28,28)),
    keras.layers.Dense(128, activation= 'relu'),
    keras.layers.Dense(10, activation= 'softmax')
])
```

```python
# compilar el modelo
model.compile(optimizer='adam',
              loss='sparse_categorical_crossentropy',
              metrics=['accuracy'])
model.summary()
```

```python
# entrenar el modelo
history = model.fit(training_images, training_labels, epochs=10)
```

```python
# graficar el historial de entrenamiento:
import pandas as pd
pd.DataFrame(history.history).plot(grid=True)
```

```python
# evaluar el modelo en el conjunto de entrenamiento
loss, accuracy = model.evaluate(training_images, training_labels)
print(f'Loss: {loss}')
print(f'Accuracy: {accuracy}')  #presicion en el conjunto de prueba

#evaluar el modelo con datos no vistos
loss, accuracy = model.evaluate(test_images, test_labels)
print(f'Loss: {loss}')
print(f'Accuracy: {accuracy}')
```
