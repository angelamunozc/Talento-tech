```python
# importar bibliotecas

import keras
from sklearn.datasets import make_classification
from sklearn.model_selection import train_test_split

# generar un conjunto de datos de clasificacion sintetico

X, y = make_classification(n_samples=1000, n_features=5, n_classes=2)

# dividir los datos en conjunto de entrenamiento y prueba

X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.2)

# convertir las etiquetas a codificacion one-hot

y_train_categorical = keras.utils.to_categorical(y_train)
y_test_categorical = keras.utils.to_categorical(y_test)

# crear una instancia de un modelo secuencial de keras

model = keras.models.Sequential()

# añadir una capa oculta con 10 neuronas y activacion RELU

model.add(keras.layers.Dense(10, activation='sigmoid'))

# añadir dos capa oculta con 5 neuronas y activacion RELU

model.add(keras.layers.Dense(5, activation='relu'))

# añadir una capa de salida con 2 neuronas (una para cada clase ) y
#activacion softmax
model.add(keras.layers.Dense(2, activation='relu'))


#compliar el modelo

model.compile(optimizer='adam', loss='categorical_crossentropy', metrics=['accuracy'])

# entrenar el modelo

model.fit(X_train, y_train_categorical, epochs=20, batch_size=40,
          validation_data=(X_test, y_test_categorical))

# evaluar el modelo en el conjunto de prueba

loss, accuracy = model.evaluate(X_test, y_test_categorical)

#impresion de resultados
print('Perdida en el conjunto de prueba:', loss)
print('Precision en el conjunto de prueba:', accuracy)
```
