```python
import keras
#cargue el conjunto de datos fashion mnist
fmnist = keras.datasets.fashion_mnist
# cargue la division de entrenamiento y rpueba del conjunto de datos fashion mnist
(training_images, training_labels), (test_images, test_labels) = fmnist.load_data()
```

```python
import numpy as np
import matplotlib.pyplot as plt
# prueba poner aqui entre 0 y 59999
index = 3
#imprime la etiqueta y la imagen
np.set_printoptions(linewidth=320)
print (f'Labels: {training_labels[index]}')
print (f'Image:\n: {training_images[index]}')
#visualiza la imagen
plt.imshow(training_images[index])
```

```python
#normalizar los valores de pixeles del tren y probar las imagenes
training_images = training_images / 255.0
test_images = test_images / 255.0
```

```python
# construir el modelo de clasificacion
model = keras.Sequential([
  keras.layers.Flatten(input_shape=(28,28)),
  keras.layers.Dense(128, activation= 'relu'),
  keras.layers.Dense(10, activation='softmax')
])
```

```python
# compilar el modelo
model.compile(optimizer='adam',
              loss='sparse_categorical_crossentropy',
              metrics=['accuracy'])
model.summary ()

#la metreica para  clasificacion se usa spaese categorica, no puede ser el error cuadratico
```

```python
# entrenar el modelo
history = model.fit(training_images, training_labels, epochs=10)
```

```python
#graficar el historial de inetrenamiento
import pandas as pd
pd.DataFrame(history.history).plot()
```

```python
#evaluar el modelo en el conjjunto de entrenamiento
loss, accuracy = model.evaluate(training_images, training_labels)
print (f'Loss: {loss}')
print (f'Accuracy: {accuracy}')

#evaluar el modelo con datos no vistos
loss, accuracy = model.evaluate(test_images, test_labels)
print (f'Loss: {loss}')
print (f'Accuracy: {accuracy}')
```

```python
from inspect import classify_class_attrs
#predict
index = 1
print (f'Label: {test_labels[index]}')
classification = model.predict(test_images[index:index+1])
print (f'Classification:\n{classification.reshape(-1,1)}')
```

# METRICAS DE CLASIFICACION

```python
from sklearn.metrics import confusion_matrix
import seaborn as sns

# Obtener las predicciones para el conjunto de prueba
test_predictions = model.predict(test_images)
test_predictions_classes = np.argmax(test_predictions, axis=1)

# Generar la matriz de confusión
cm = confusion_matrix(test_labels, test_predictions_classes)

# Graficar la matriz de confusión
plt.figure(figsize=(10, 8))
sns.heatmap(cm, annot=True, fmt='d', cmap='Blues')
plt.xlabel('Etiqueta Predicha')
plt.ylabel('Etiqueta Verdadera')
plt.title('Matriz de Confusión')
plt.show()
```
